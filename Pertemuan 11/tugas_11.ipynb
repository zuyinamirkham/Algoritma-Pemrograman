Buatlah program sistem manajemen perpustakaan sederhana (menambahkan dan
menampilkan buku baru ke dalam sistem) menggunakan konsep multi-keys dan nesting
dictionary dengan format
- KEYS
- JUDUL BUKU
- PENULIS
- TAHUN TERBIT
- TANGGAL TERAKHIR PINJAM
- STATUS (TERSEDIA / TIDAK)

```python
import datetime
import string
import random

perpustakaan_template = {
    "judul" : "judul buku",
    "penulis" : "nama",
    "tahun_terbit" : 0000,
    "pinjam" : datetime.datetime(1111,1,11),
    "status" : "tersedia/tidak"
}

data_perpustakaan = {}
while True:
  print(f"{'MANAGEMEN PERPUSTAKAAN':^40}")
  print(f"{'DATA BUKU':^40}")
  print("-"*40)

  buku = dict.fromkeys(perpustakaan_template.keys())
  buku['judul'] = input("Judul Buku: ")
  buku['penulis'] = input("Nama Penulis: ")
  buku['tahun_terbit'] = int(input("Tahun_Terbit: "))
  TAHUN_PINJAM = int(input("Tahun Terakhir Pinjam (YYYY): "))
  BULAN_PINJAM = int(input("Bulan Terakhir Pinjam (1-12): "))
  TANGGAL_PINJAM = int(input("Tanggal Terakhir Pinjam (1-31): "))
  buku['pinjam'] = datetime.datetime(TAHUN_PINJAM,BULAN_PINJAM,TANGGAL_PINJAM)
  buku['status'] = input("Apakah buku tersedia? (y/n): ").lower()
  if buku ['status'] == "y":
      buku["status"] = True
  else:
      buku["status"] = False
  print("\nData buku berhasil ditambahkan:")
  print(buku)

  KEY = ''.join((random.choice(string.ascii_uppercase) for i in range(6)))
  data_perpustakaan.update({KEY:buku})
  print(f"\n{'KEY':<10} {'judul':<20} {'penulis':<13} {'tahun_terbit':<10} {'Tanggal terakhir pinjam':<25} {'status':<10}")
  print('-'*95)
  for KEY, buku in data_perpustakaan.items():
    JUDUL = buku['judul']
    PENULIS = buku['penulis']
    TAHUN_TERBIT = buku['tahun_terbit']
    PINJAM = buku['pinjam'].strftime("%x")
    STATUS = "Tersedia" if buku["status"] else "Dipinjam"
    print(f"{KEY:<10} {JUDUL:<20} {PENULIS:<13} {TAHUN_TERBIT:^10} {PINJAM:^25} {STATUS:^10}")
  print("\n")
  is_done = input("Mau tambah lagi bro (y/n)? ")
  if is_done == "n":
    break

print("\nAkhir dari program, terima kasih")
```
