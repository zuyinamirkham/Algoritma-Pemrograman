A.Luas Persegi Panjang

```python

persegi1 = float(input("masukkan angka untuk panjang: "))
persegi2 = float(input("masukkan angka untuk lebar: "))

def bangun(persegi1,persegi2):
  return persegi1 * persegi2

luas = bangun(persegi1,persegi2)
print (f"luas persegi panjang tersebut adalah {luas}")
```

B.Volume Balok

```python

panjang = float(input("masukkan angka untuk panjang: "))
lebar = float(input("masukkan angka untuk lebar: "))
tinggi = float(input("masukkan angka untuk tinggi: "))

def volume(panjang,lebar,tinggi):
  return panjang * lebar * tinggi

hasil = volume(panjang,lebar,tinggi)
print(f"volume balok tersebut adalah {hasil}")
```

C.Luas Lingkaran


```python
jari2_input = float(input("masukkan angka untuk jari-jari: "))

def luas(phi = 3.14, jari_jari = 0):
  return phi * jari_jari ** 2

hasil = luas(phi = 3.14, jari_jari = jari2_input)
print(f"luas lingkaran tersebut adalah {hasil}")
```

D.Volume Bola

```python
jari2_input = float(input("masukkan angka untuk jari-jari: "))

def volume(phi = 3.14,jari_jari = 0):
  return 4/3 * phi * jari_jari ** 3

hasil = volume(phi = 3.14,jari_jari = jari2_input)
print(f"volume bola tersebut adalah {hasil}")
```
