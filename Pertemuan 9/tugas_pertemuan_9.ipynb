```python
# nama : zuyina m. irkham
# npm  : 2505060041

# 1.
buah = ["apel","mangga","jeruk","anggur","pisang"]
print(buah)

# a.tambahkan "semangka" di akhir list
buah.append("semangka") # tanpa tambah index ke berapa
print(f"a.data buah setelah di tambah semangka: {buah}")

# b.sisipkan "durian" di antara "jeruk" dan "anggur"
buah.insert(3,"durian")
print(f"b.data buah setelah disisipkan durian: {buah}")

# c.hapus "mangga" dari list
buah.remove("mangga")
print(f"c.data buah setelah mangga di hapus: {buah}")

# d.ubah "pisang" menjadi "nanas"
buah[4] = "nanas"
print(f"d.data buah setelah pisang di ubah menjadi nanas: {buah}")

# e.tampilkan 3 buah pertama
print(f"e.3 buah pertama adalah {buah[0]},{buah[1]},{buah[2]}")
```
