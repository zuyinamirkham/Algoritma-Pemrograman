2.dictionary

```python
# buat dictionary keranjang
keranjang = {
    "apel" : 5000,
    "pisang" : 3000,
    "jeruk" : 7000,
    "anggur" : 15000,
}
print(keranjang)
# membuat tampilan list harga
total = 0 # untuk menampung
for buah,harga in keranjang.items():
  print(f"Harga {buah}: Rp.{harga}")
  total += harga # menghitung total semua barang
print(f"harga total: Rp. {total}")



```
